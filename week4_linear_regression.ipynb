```python
# Week 4: Simple Linear Regression on Heart Data
"""
**Question:** Does maximum heart rate (MaxHR) fall as age rises?

**Data:** Kaggle Heart Failure Prediction dataset (918 patients).

**Method:** Simple linear regression with scikit-learn, MaxHR as the target and Age as the predictor.

**Key result:** For each extra year of age, MaxHR falls by about ___ bpm. R² = ___, so Age explains about ___% of the variation."""
```

```python
#data set loading
import pandas as pd
df=pd.read_csv("heart.csv")
```

```python
#First Few top Rows
df.head()
```

```python
#Size of data
df.shape
```

```python
#Fitting simple linear regression line 
from sklearn.linear_model import LinearRegression
x1=df[["Age"]]
y1=df["MaxHR"]
lin=LinearRegression()
lin.fit(x1,y1)
print("Intercept:",lin.intercept_)
print("slope:",lin.coef_[0])
print("R_squared:",lin.score(x1,y1))
```

```python
#Visualization of regression line
import matplotlib.pyplot as plt
plt.scatter(df['Age'],df['MaxHR'],alpha=0.4)
plt.plot(df['Age'],lin.predict(x1),color='red')
plt.xlabel('Age')
plt.ylabel('MaxHR')
plt.show()
```

```python
#prediction based on regression 
lin.predict(pd.DataFrame({'Age':[40,60]}))
```

```python
#summary 
"""A simple linear regression of MaxHR on Age gave a slope of about -1 (MaxHR falls roughly 1 bpm per year of age)
and R² = 0.1459, so Age explains about 14.6% of the variation in MaxHR. The intercept is the line’s value at Age 0, 
which is far outside the data (youngest patients are in their late 20s), so it is just a starting point for the line and has no real meaning."""
```

```python
# Limitations
"""- R² is low, so Age alone is a weak predictor of MaxHR.
- The intercept (value at age 0) lies far outside the data (youngest patients are in their late 20s),
  so it has no real meaning.
- The relationship is correlation only, not proof that age causes the drop."""
```
